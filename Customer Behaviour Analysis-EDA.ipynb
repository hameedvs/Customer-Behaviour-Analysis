```python
correct_mapping = {
    #clothing
    "T-shirt":"Clothing","Shirt":"Clothing","Shorts":"Clothing","Hoodie":"Clothing",
    "Pants":"Clothing","Socks":"Clothing","Jeans":"Clothing","Blouse":"Clothing",
    "Skirt":"Clothing","Sweater":"Clothing","Dress":"Clothing",

    #Accessories
    "Sunglasses":"Accessories","Gloves":"Accessories","Jewelry":"Accessories",
    "Hat":"Accessories","Handbag":"Accessories","Backpack":"Accessories",
    "Belt":"Accessories","Bag":"Accessories","Scarf":"Accessories",

    #Electronics
    "Laptop":"Electronics","Phone":"Electronics","Watch":"Electronics","Headphones":"Electronics",

    #Footwear
    "Shoes":"Footwear","Sandals":"Footwear","Sneakers":"Footwear","Boots":"Footwear",

    #Outerwear
    "Coat":"Outerwear","Jacket":"Outerwear"
    
}
```

## Treating Null Values

```python
df.groupby('Category')['Size'].unique()
```

```python
#Treating null values in Electronics
df.loc[df['Category']=='Electronics','Size'] = "Not Applicable"
```

```python
#Treating null values in Accessories
df.loc[df['Category']=='Accessories','Size'] = "Free Size"
```

```python
df.groupby('Category')['Size'].unique()
```

```python
#Treating null values in Footwear
df.loc[df['Category']=='Footwear','Size'] = "Not Available"
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.loc[(df['Category'] == 'Clothing') & (df['Size'].isnull()),'Size'] = df[df['Category'] == "Clothing"]['Size'].mode()[0]
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.isnull().sum()
```

## Review Rating

```python
df['Review Rating'].mean()
```

```python
df['Review Rating'].median()
```

```python
df['Review Rating'].mode()
```

```python
df.groupby("Item Purchased")["Review Rating"].mean()
```

```python
# filling the null value of review ratings with the mean of each item
df['Review Rating']=df['Review Rating'].fillna(df.groupby("Item Purchased")['Review Rating'].transform('mean'))
```

```python
df.isnull().sum()
```

```python
df
```

```python
df['Review Rating']=df['Review Rating'].round(2)
```

```python
df.head()
```

## Previous Purchases

```python
df['Previous Purchases']=df['Previous Purchases'].fillna(0)
```

```python
df.isnull().sum()
```

## Purchase Amount (USD)

```python
df.groupby("Item Purchased")["Purchase Amount (USD)"].mean()
```

```python
# filling the null value of purchase amount with the mean of each purchase
df['Purchase Amount (USD)']=df['Purchase Amount (USD)'].fillna(df.groupby("Item Purchased")['Purchase Amount (USD)'].transform('mean'))
```
